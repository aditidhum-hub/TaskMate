## 1. Setup Environment and Path

```python
import os
import sys

# Ensure project root is in sys.path
project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from backend.app.core.config import get_settings
from backend.app.services.llm_service import LLMService, LLMAuthenticationError

settings = get_settings()
print(f"LLM Provider:  {settings.LLM_PROVIDER}")
print(f"LLM Model:     {settings.LLM_MODEL}")
print(f"Base URL:      {settings.effective_base_url}")
print(f"API Key Configured: {'Yes (Length: ' + str(len(settings.effective_api_key)) + ')' if settings.effective_api_key else 'No (Missing)'}")
```

## 2. Initialize LLMService

```python
service = LLMService(settings=settings)
print(f"Service Model:    {service.model}")
print(f"Service Base URL: {service.base_url}")
print("✅ LLMService initialized successfully!")
```

## 3. Test Direct Chat Completion (Conversational Prompt)

> Note: Requires a valid `NVIDIA_API_KEY` or `LLM_API_KEY` in environment or `.env`.

```python
messages = [
    {"role": "system", "content": "You are TaskMate, an AI assistant for personal task management."},
    {"role": "user", "content": "Hello! Who are you and how can you help me?"}
]

try:
    response = service.chat_completion(messages=messages)
    print("--- Response from Nemotron ---")
    print(response.content)
    print("------------------------------")
    print(f"Finish Reason: {response.finish_reason}")
    print("✅ Chat completion succeeded!")
except LLMAuthenticationError as err:
    print(f"⚠️ Authentication error (expected if NVIDIA_API_KEY is not set): {err}")
except Exception as err:
    print(f"⚠️ Service error: {err}")
```
