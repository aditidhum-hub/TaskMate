## 3. Verify Server-Side Token Verification (`backend.app.core.security`)

```python
from backend.app.core.security import verify_firebase_token

# Test with valid token pattern
token = "mock-valid-token-user_test_987"
claims = verify_firebase_token(token)

print(f"Decoded Token Claims: {claims}")
print(f"Authenticated UID:    {claims['uid']}")

assert claims["uid"] == "user_test_987"
print("✅ User identity successfully derived from verified token!")
```

## 4. Verify Rejection of Expired and Malformed Tokens

```python
from fastapi import HTTPException

for invalid_token in ["mock-expired-token", "mock-invalid-token", ""]:
    try:
        verify_firebase_token(invalid_token)
        print(f"❌ Failed: {invalid_token} should have raised HTTPException")
    except HTTPException as e:
        assert e.status_code == 401
        print(f"✅ Correctly rejected invalid token ({invalid_token}): HTTP {e.status_code} - {e.detail}")
```
